```python
from urllib.parse import urlparse

# Company / Executive records
company_df = dataset1[
    dataset1["organization_type"] == "Company/Executive"
].copy()

# Extract source domain
company_df["source_domain"] = company_df["source_url"].apply(
    lambda x: urlparse(x).netloc.lower().replace("www.", "")
)

# Normalize official domain
company_df["official_domain_clean"] = (
    company_df["official_domain"]
    .str.lower()
    .str.replace("www.", "", regex=False)
)

# Normalize source domain
company_df["source_domain_clean"] = (
    company_df["source_domain"]
    .str.lower()
    .str.replace("www.", "", regex=False)
)

# Exact domain match
company_df["domain_match"] = (
    company_df["official_domain_clean"]
    == company_df["source_domain_clean"]
)

# Display all domain relationships
company_domain_check = (
    company_df[
        [
            "organization",
            "official_domain_clean",
            "source_domain_clean",
            "domain_match"
        ]
    ]
    .drop_duplicates()
    .sort_values(
        ["domain_match", "organization"],
        ascending=[True, True]
    )
)

print("Total Company/Executive records:", len(company_df))
print(
    "Organizations:",
    company_df["organization"].nunique()
)
print(
    "Exact domain matches:",
    company_domain_check["domain_match"].sum()
)
print(
    "Domain mismatches:",
    (~company_domain_check["domain_match"]).sum()
)

display(company_domain_check.to_string(index=False))
```

```python
import re

def normalize_name(name):
    name = re.sub(r"\b(Dr|Mr|Mrs|Ms|Smt|Shri|Sri|Er)\.?\b", "", str(name), flags=re.I)
    name = re.sub(r"\b(IAS|OAS|OFS|OSS|SAG|SS)\b", "", name, flags=re.I)
    return re.sub(r"[^a-z0-9]", "", name.lower())

def email_local(email):
    return str(email).lower().split("@")[0]

dataset1["name_clean"] = dataset1["name"].apply(normalize_name)
dataset1["email_local"] = dataset1["official_email"].apply(email_local)

# Check whether at least part of the person's name appears in the email
def name_email_match(row):
    name = row["name_clean"]
    email = row["email_local"].replace(".", "").replace("_", "").replace("-", "")
    
    if len(name) < 4:
        return True
    
    # first name / last name tokens
    original_name = re.sub(
        r"\b(Dr|Mr|Mrs|Ms|Smt|Shri|Sri|Er|IAS|OAS|OFS|OSS|SAG|SS)\.?\b",
        "",
        str(row["name"]),
        flags=re.I
    )
    
    tokens = [
        re.sub(r"[^a-z]", "", x.lower())
        for x in original_name.split()
    ]
    tokens = [x for x in tokens if len(x) >= 4]

    return any(token in email for token in tokens)

dataset1["name_email_match"] = dataset1.apply(
    name_email_match,
    axis=1
)

print("Name-email match:", dataset1["name_email_match"].sum())
print("Name-email flagged:", (~dataset1["name_email_match"]).sum())

display(
    dataset1[~dataset1["name_email_match"]][
        [
            "identity_id",
            "name",
            "organization",
            "official_email",
            "official_domain"
        ]
    ].to_string(index=False)
)
```

```python
import re
import pandas as pd

# -----------------------------------------
# 1. Extract email parts
# -----------------------------------------

def clean_name(name):
    name = re.sub(
        r"\b(Dr|Mr|Mrs|Ms|Smt|Shri|Sri|Er|IAS|OAS|OFS|OSS|SAG|SS)\.?\b",
        "",
        str(name),
        flags=re.I
    )
    return re.findall(r"[a-zA-Z]+", name.lower())


def email_local(email):
    return str(email).split("@")[0].lower()


def email_domain(email):
    parts = str(email).lower().split("@")
    return parts[1] if len(parts) == 2 else ""


# -----------------------------------------
# 2. Name ↔ Email check
# -----------------------------------------

def check_name_match(row):

    parts = clean_name(row["name"])
    local = re.sub(r"[^a-z]", "", email_local(row["official_email"]))

    # Remove very short initials
    meaningful_parts = [p for p in parts if len(p) >= 3]

    if not meaningful_parts:
        return "UNKNOWN"

    if any(part in local for part in meaningful_parts):
        return "MATCH"

    return "NO_MATCH"


# -----------------------------------------
# 3. Domain ↔ Official Domain check
# -----------------------------------------

def check_domain_match(row):

    email_dom = email_domain(row["official_email"])
    official_dom = str(row["official_domain"]).lower().strip()

    if not email_dom or not official_dom:
        return "INVALID"

    if email_dom == official_dom:
        return "MATCH"

    # Allow subdomains
    if email_dom.endswith("." + official_dom):
        return "SUBDOMAIN"

    if official_dom.endswith("." + email_dom):
        return "PARENT_DOMAIN"

    return "MISMATCH"


# -----------------------------------------
# 4. Detect public email providers
# -----------------------------------------

public_domains = {
    "gmail.com",
    "yahoo.com",
    "yahoo.co.in",
    "outlook.com",
    "hotmail.com",
    "rediffmail.com"
}


def check_public_email(row):
    return email_domain(row["official_email"]) in public_domains


# -----------------------------------------
# 5. Detect obvious role/shared addresses
# -----------------------------------------

role_patterns = [
    "director", "dean", "registrar", "secretary", "secy",
    "chairman", "chairperson", "cmo", "finance", "finsec",
    "admin", "office", "department", "dept", "investor",
    "institutional", "info", "support", "help", "rti",
    "media", "lab", "works", "tourism", "energy", "coop",
    "education", "health", "welfare", "development",
    "despatch", "pmu", "mo.", "ad.", "dd.", "as.",
    "js.", "dir-", "ce", "ee", "ae"
]


def check_role_email(row):

    local = email_local(row["official_email"])

    for pattern in role_patterns:
        if pattern in local:
            return True

    return False


# -----------------------------------------
# 6. Create full 500-record audit
# -----------------------------------------

email_audit = dataset1.copy()

email_audit["email_domain"] = email_audit.apply(
    email_domain, axis=1
)

email_audit["name_email_check"] = email_audit.apply(
    check_name_match, axis=1
)

email_audit["domain_check"] = email_audit.apply(
    check_domain_match, axis=1
)

email_audit["public_email"] = email_audit.apply(
    check_public_email, axis=1
)

email_audit["role_or_shared_email"] = email_audit.apply(
    check_role_email, axis=1
)


# -----------------------------------------
# 7. Duplicate email count
# -----------------------------------------

email_counts = (
    email_audit["official_email"]
    .str.lower()
    .value_counts()
)

email_audit["email_usage_count"] = (
    email_audit["official_email"]
    .str.lower()
    .map(email_counts)
)


# -----------------------------------------
# 8. Overall review category
# -----------------------------------------

def classify_email(row):

    # Clearly invalid domain
    if row["domain_check"] == "INVALID":
        return "INVALID"

    # Domain does not match recorded official domain
    if row["domain_check"] == "MISMATCH":
        return "DOMAIN_REVIEW"

    # Shared/role address
    if row["role_or_shared_email"]:
        return "ROLE_OR_SHARED_REVIEW"

    # Public provider
    if row["public_email"]:
        return "PUBLIC_PROVIDER_REVIEW"

    # Individual email doesn't relate to name
    if row["name_email_check"] == "NO_MATCH":
        return "NAME_REVIEW"

    return "PASS"


email_audit["email_status"] = email_audit.apply(
    classify_email,
    axis=1
)


# -----------------------------------------
# 9. Summary
# -----------------------------------------

print("Total records:", len(email_audit))
print("\nEmail status:")
print(email_audit["email_status"].value_counts())

print("\nName ↔ Email:")
print(email_audit["name_email_check"].value_counts())

print("\nDomain check:")
print(email_audit["domain_check"].value_counts())

print("\nPublic provider emails:",
      email_audit["public_email"].sum())

print("\nShared/duplicate emails:",
      (email_audit["email_usage_count"] > 1).sum())


# -----------------------------------------
# 10. Show ALL records needing review
# -----------------------------------------

review = email_audit[
    email_audit["email_status"] != "PASS"
].copy()

display(
    review[
        [
            "identity_id",
            "name",
            "organization",
            "official_email",
            "official_domain",
            "email_domain",
            "name_email_check",
            "domain_check",
            "public_email",
            "role_or_shared_email",
            "email_usage_count",
            "email_status"
        ]
    ].to_string(index=False)
)
```

```python
# CHECK 23A — NORMALIZED SELF-SENDER NAME CONSISTENCY

import re

def normalize_person_name(name):
    name = str(name).strip().lower()

    # Remove common honorifics/titles
    titles = [
        "dr", "mr", "mrs", "ms", "smt", "shri", "sri",
        "prof", "professor"
    ]

    for title in titles:
        name = re.sub(rf"\b{title}\.?\b", " ", name)

    # Remove common professional/designation suffixes
    suffixes = [
        "ias", "ips", "ifs", "oas", "osjs", "irs",
        "o a s", "o.a.s", "o.a.s.", "sag", "osd"
    ]

    for suffix in suffixes:
        name = re.sub(rf"\b{re.escape(suffix)}\b", " ", name)

    # Remove punctuation
    name = re.sub(r"[^a-z0-9\s]", " ", name)

    # Normalize spaces
    name = re.sub(r"\s+", " ", name).strip()

    return name


# Official-email records
email_self["identity_name_normalized"] = (
    email_self["identity_name"].apply(normalize_person_name)
)

email_self["sender_name_normalized"] = (
    email_self["sender_name"].apply(normalize_person_name)
)

# Compare normalized names
normalized_name_mismatch = (
    email_self["identity_name_normalized"]
    !=
    email_self["sender_name_normalized"]
)

print("Official-email self-sender records:", len(email_self))
print(
    "Normalized name mismatches:",
    normalized_name_mismatch.sum()
)

if normalized_name_mismatch.sum() > 0:
    print("\nPotential genuine name mismatches:")
    print(
        email_self.loc[
            normalized_name_mismatch,
            [
                "identity_id",
                "identity_name",
                "sender_name",
                "sender_email",
                "identity_name_normalized",
                "sender_name_normalized"
            ]
        ]
        .drop_duplicates()
        .to_string(index=False)
    )
else:
    print("\nPASS — Names match after removing titles/designations.")
```

```python
# CHECK 23B — IMPROVED NAME NORMALIZATION

def normalize_person_name_v2(name):
    name = str(name).strip().lower()

    # Remove anything inside parentheses
    name = re.sub(r"\([^)]*\)", " ", name)

    # Remove common titles/designations
    tokens_to_remove = [
        "dr", "mr", "mrs", "ms", "smt", "shri", "sri",
        "prof", "professor", "er",
        "ias", "ips", "ifs", "oas", "oss", "osjs",
        "ofs", "irs", "adg", "ss", "sag", "osd"
    ]

    for token in tokens_to_remove:
        name = re.sub(
            rf"\b{re.escape(token)}\.?\b",
            " ",
            name
        )

    # Remove punctuation
    name = re.sub(r"[^a-z0-9\s]", " ", name)

    # Normalize spaces
    name = re.sub(r"\s+", " ", name).strip()

    return name


email_self["identity_name_normalized_v2"] = (
    email_self["identity_name"].apply(normalize_person_name_v2)
)

email_self["sender_name_normalized_v2"] = (
    email_self["sender_name"].apply(normalize_person_name_v2)
)

name_mismatch_v2 = (
    email_self["identity_name_normalized_v2"]
    !=
    email_self["sender_name_normalized_v2"]
)

print("Official-email self-sender records:", len(email_self))
print("Remaining normalized name mismatches:", name_mismatch_v2.sum())

if name_mismatch_v2.sum() > 0:
    print("\nRemaining potential mismatches:")
    print(
        email_self.loc[
            name_mismatch_v2,
            [
                "identity_id",
                "identity_name",
                "sender_name",
                "sender_email",
                "identity_name_normalized_v2",
                "sender_name_normalized_v2"
            ]
        ]
        .drop_duplicates()
        .to_string(index=False)
    )
else:
    print("\nPASS — All official-email self-sender names match after normalization.")
```
